```python
from pathlib import Path
import numpy as np
import pandas as pd
import os
```

```python
# Open the directory for the embeddings
model_directories = {
    'perch_bird': Path(f'embeddings/2026-09-07_08-34___perch_bird-ssw_clips_mixed'),
    'birdmae': Path("/home/skcragg/git_local/fa2026-workshop/embeddings/2026-09-07_08-35___birdmae-ssw_clips_mixed"),
    'perch_v2' : Path("/home/skcragg/git_local/fa2026-workshop/embeddings/2026-09-07_08-37___perch_v2-ssw_clips_mixed")} # set this as embedding network - as a list maybe and iterate?

df = pd.DataFrame({
                    "filename": pd.Series(dtype="str"),
                    "orig_id": pd.Series(dtype="str"),
                    "species": pd.Series(dtype="str"),
                    "clean" : pd.Series(dtype="bool"),
                    "snr": pd.Series(dtype="float64"),
                    "embedding": pd.Series(dtype="object"),
                    "model" : pd.Series(dtype="str")
                })

# Per folder all the embeddings per network will be here
# File name {class_name}_{tag}_{snr value}.npy

for model_name, directory in model_directories.items():
    for filename in Path(directory).rglob('*.npy'):
        #print(filename)
        # Parse name
        #print(Path(filename).stem.replace(model_name, '').split('_'))
        name_parts = Path(filename).stem.replace(model_name, '').split('_')

        clean = False
        species = name_parts[-4]
        snr = name_parts[-2]
        orig_id = name_parts[-3]
        if snr == 'clean':
            clean = True
            snr = None
        else:
            snr = float(snr)

        embedding = np.load(Path(filename))

        df.loc[len(df)] = {
            "filename": Path(filename).stem,
            "orig_id": None, # PARSE THIS
            "species": species,
            "clean": clean,
            "snr": snr,
            "embedding": embedding,
            "model": model_name,
        }

# Check dataframe has been correctly populated
df.head()
```

```python
# Compute embedding distance to the clean signal
# silhouette_score

distances = pd.Series(index=df.index, dtype="float64")

for (model, species, tag), group in df.groupby(["model", "species", "tag"]):
        clean_rows = group[group["clean"]]

        if clean_rows.empty:
            print(f"No clean reference for {model}/{species}/{tag}, leaving distances as NaN")
            continue
        if len(clean_rows) > 1:
            print(f"Warning: multiple clean refs for {model}/{species}/{tag}, using first")

```
