```python
from pathlib import Path
import numpy as np
import pandas as pd
import os
```

```python
# Open the directory for the embeddings
model_directories = {'model_name': Path(f'')} # set this as embedding network - as a list maybe and iterate?

df = pd.DataFrame({
                    "filename": pd.Series(dtype="str"),
                    "species": pd.Series(dtype="str"),
                    "clean_id" : pd.Series(dtype="str"),
                    "snr": pd.Series(dtype="float64"),
                    "embedding": pd.Series(dtype="object"),
                    "model" : pd.Series(dtype="str")
                })

# Per folder all the embeddings per network will be here
# File name {class_name}_{tag}_{snr value}.npy

for model_name, directory in model_directories:
    for filename in os.listdir(directory):
        # Parse name
        name_parts = Path(filename).stem().split('_')

        species = name_parts[0]
        snr = name_parts[]

        embedding = np.load(Path(filename))

        df.loc[len(df)] = {
            "filename": Path(filename).stem(),
            "species": species,
            "clean_id": None,
            "snr": snr,
            "embedding": embedding,
            "model": model_name,
        }
```
